# 📊 Análisis del dataset `matching.json`

Análisis exploratorio básico del fichero de ejercicios de tipo *matching*.


```python
import json
from collections import defaultdict, Counter

with open("../../data/dev_matching/matching.json") as f:
    data = json.load(f)

exams = data["exams"]
print(f"Total de exámenes cargados: {len(exams)}")
```

## 1. Visión general

```python
exercises = []
for exam in exams:
    for ex in exam["exercises"]:
        ex["_level"] = exam["level"]
        ex["_examId"] = exam["examId"]
        exercises.append(ex)

print(f"Total de ejercicios:  {len(exercises)}")
print(f"Total de preguntas:   {sum(e['num-questions'] for e in exercises)}")
print(f"Niveles presentes:    {sorted(set(e['_level'] for e in exercises))}")
```

## 2. Ejercicios por nivel

```python
from collections import Counter

nivel_count   = Counter(e["_level"] for e in exercises)
nivel_pregs   = defaultdict(int)
for e in exercises:
    nivel_pregs[e["_level"]] += e["num-questions"]

print(f"{'Nivel':<8} {'Ejercicios':>12} {'Preguntas':>12}")
print("-" * 34)
for nivel in sorted(nivel_count):
    print(f"{nivel:<8} {nivel_count[nivel]:>12} {nivel_pregs[nivel]:>12}")
```

## 3. Ejercicios con imágenes

```python
def has_images(ex):
    s1_imgs = [x for x in ex["exercise"]["set1"] if x.get("image-path", "")]
    s2_imgs = [x for x in ex["exercise"]["set2"] if x.get("image-path", "")]
    return bool(s1_imgs or s2_imgs), len(s1_imgs), len(s2_imgs)

print(f"{'ExerciseID':<28} {'Nivel':<6} {'¿Imágenes?':<12} {'Imgs set1':>10} {'Imgs set2':>10}")
print("-" * 70)
total_con_imgs = 0
for ex in exercises:
    tiene, n1, n2 = has_images(ex)
    marca = "✅ Sí" if tiene else "❌ No"
    if tiene:
        total_con_imgs += 1
    print(f"{ex['exerciseID']:<28} {ex['_level']:<6} {marca:<12} {n1:>10} {n2:>10}")

print(f"\nEjercicios con imágenes: {total_con_imgs} / {len(exercises)}")
```

## 4. Distribución del número de preguntas

```python
dist = Counter(e["num-questions"] for e in exercises)
print(f"{'num-questions':>15} {'frecuencia':>12}")
print("-" * 29)
for k in sorted(dist):
    print(f"{k:>15} {dist[k]:>12}")
```

## 5. Resumen por examen

```python
print(f"{'examId':<22} {'Nivel':<6} {'Ejercicios':>12} {'Preguntas':>10}")
print("-" * 54)
for exam in exams:
    n_ex   = len(exam["exercises"])
    n_preg = sum(e["num-questions"] for e in exam["exercises"])
    print(f"{exam['examId']:<22} {exam['level']:<6} {n_ex:>12} {n_preg:>10}")
```
